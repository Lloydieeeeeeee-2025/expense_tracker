```python
%sql
CREATE OR REFRESH STREAMING TABLE expense_tracker_catalog.expense_tracker_schema.expense_silver AS
SELECT
    CAST(expense_id AS STRING) AS expense_id,
    CAST(expense_date AS STRING) AS expense_date,
    CAST(amount AS DOUBLE) AS expense_amount,
    CAST(payment_method AS STRING) AS payment_method,
    description,
    CAST(category_id AS STRING) AS category_id,
    CAST(user_id AS INT) AS user_id,
    CAST(created_at AS TIMESTAMP) AS created_at
FROM STREAM(expense_tracker_catalog.expense_tracker_schema.expense)
WHERE NULLIF(TRIM(expense_id), '') IS NOT NULL
    AND NULLIF(TRIM(user_id), '') IS NOT NULL
    AND NULLIF(TRIM(category_id), '') IS NOT NULL
    AND NULLIF(TRIM(expense_date), '') IS NOT NULL
    AND NULLIF(TRIM(amount), '') IS NOT NULL
    AND NULLIF(TRIM(payment_method), '') IS NOT NULL
    AND NULLIF(TRIM(created_at), '') IS NOT NULL
```
